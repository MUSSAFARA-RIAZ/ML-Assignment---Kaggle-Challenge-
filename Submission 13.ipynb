## Data Cleaning

```python
def clean_text(series):
    result = series.astype('string').str.strip().str.upper()
    result = result.replace({'?': pd.NA, 'UNKNOWN': pd.NA, 'U': pd.NA,
                             'NOT AVAILABLE': pd.NA, 'NOT MAPPED': pd.NA, '': pd.NA})
    return result.astype(object).where(result.notna(), np.nan)

for table, columns in [
    (patients, ['insurance_type', 'race_ethnicity', 'sex', 'language', 'marital_status']),
    (encounters, ['admission_type', 'admission_source', 'service_line', 'discharge_disposition', 'attending_specialty']),
    (hospitals, ['urban_rural', 'ownership']),
]:
    for column in columns:
        table[column] = clean_text(table[column])
encounters['admit_dt'] = pd.to_datetime(encounters.admit_dt, format='mixed', errors='coerce')
encounters['discharge_dt'] = pd.to_datetime(encounters.discharge_dt, format='mixed', errors='coerce')
assert encounters.admit_dt.notna().all() and encounters.discharge_dt.ge(encounters.admit_dt).all()
patients.loc[patients.birth_year.le(1900), 'birth_year'] = np.nan
patients.loc[~patients.deprivation_index.between(0, 100), 'deprivation_index'] = np.nan
planning['home_health_referral'] = clean_text(planning.home_health_referral).map({'Y': 1., 'N': 0.})
planning['education_done'] = clean_text(planning.education_done).map({'YES': 1., 'NO': 0.})

admit_by_stay = encounters.set_index('encounter_id').admit_dt
discharge_by_stay = encounters.set_index('encounter_id').discharge_dt

def within_stay(table, column, fmt, date_only=False):
    when = pd.to_datetime(table[column], format=fmt, errors='coerce')
    start = table.encounter_id.map(admit_by_stay)
    end = table.encounter_id.map(discharge_by_stay)
    if date_only:
        when, start, end = when.dt.normalize(), start.dt.normalize(), end.dt.normalize()
    keep = when.ge(start) & when.le(end)
    quality[column + '_outside_stay_or_invalid'] = int((~keep).sum())
    result = table.loc[keep].copy()
    result[column] = when.loc[keep]
    return result

labs = within_stay(labs, 'collected_dt', '%Y-%m-%d %H:%M:%S')
vitals = within_stay(vitals, 'charttime', '%m/%d/%Y %H:%M')
meds = within_stay(meds, 'start_dt', '%Y%m%d', date_only=True)
procedures = within_stay(procedures, 'proc_dt', '%d-%b-%Y', date_only=True)
patient_admit = encounters.set_index('patient_id').admit_dt.dt.normalize()
prior['days_before'] = (prior.patient_id.map(patient_admit) - pd.to_datetime(prior.visit_dt, errors='coerce')).dt.days
quality['prior_not_before_admission'] = int((~prior.days_before.gt(0)).sum())
prior = prior.loc[prior.days_before.gt(0)].copy()
```

Exact duplicate rows are removed, category spelling is normalized, and dates are parsed. Events outside the stay are excluded. Unknown values remain unknown.

## Lab Units and Repair

```python
conversions = {
    'Albumin': {'g/dL': 1, 'g/L': .1}, 'Hemoglobin': {'g/dL': 1, 'g/L': .1},
    'Creatinine': {'mg/dL': 1, 'umol/L': 1 / 88.4}, 'Glucose': {'mg/dL': 1, 'mmol/L': 18.0182},
    'Sodium': {'mmol/L': 1}, 'Potassium': {'mmol/L': 1}, 'Bun': {'mg/dL': 1},
    'Bnp': {'pg/mL': 1}, 'Hba1C': {'%': 1}, 'Inr': {'ratio': 1},
    'Lactate': {'mmol/L': 1}, 'Troponin_I': {'ng/mL': 1}, 'Wbc': {'10^3/uL': 1},
}
limits = {'Sodium': (80, 200), 'Potassium': (1, 10), 'Glucose': (10, 1500),
          'Creatinine': (.05, 30), 'Bun': (0, 300), 'Hemoglobin': (1, 30),
          'Wbc': (.1, 200), 'Albumin': (.1, 10), 'Lactate': (0, 40),
          'Troponin_I': (0, 100), 'Bnp': (0, 100000), 'Hba1C': (2, 25), 'Inr': (.1, 20)}

# Repair 1: infer a missing unit from the size of the value
labs['value'] = pd.to_numeric(labs.value, errors='coerce')
no_unit = labs.unit.isna()
quality['lab_units_missing_before_repair'] = int(no_unit.sum())
single_unit = {name: list(units)[0] for name, units in conversions.items() if len(units) == 1}
labs.loc[no_unit, 'unit'] = labs.loc[no_unit, 'lab_name'].map(single_unit)
size_rules = {'Albumin': ('g/L', 'g/dL', 10), 'Hemoglobin': ('g/L', 'g/dL', 30),
              'Creatinine': ('umol/L', 'mg/dL', 25), 'Glucose': ('mg/dL', 'mmol/L', 35)}
for name, (large_unit, small_unit, cut) in size_rules.items():
    rows = no_unit & labs.lab_name.eq(name)
    labs.loc[rows, 'unit'] = np.where(labs.loc[rows, 'value'] > cut, large_unit, small_unit)

labs['value_std'] = np.nan
for name, units in conversions.items():
    rows = labs.lab_name.eq(name)
    labs.loc[rows, 'value_std'] = labs.loc[rows, 'value'] * labs.loc[rows, 'unit'].map(units)
    low, high = limits[name]
    # Repair 2: a value ten times too large becomes plausible after dividing by ten
    decimal_error = rows & labs.value_std.gt(high) & (labs.value_std / 10).between(low, high)
    quality[name + '_decimal_errors_fixed'] = int(decimal_error.sum())
    labs.loc[decimal_error, 'value_std'] = labs.loc[decimal_error, 'value_std'] / 10
    invalid = rows & labs.value_std.notna() & ~labs.value_std.between(low, high)
    quality[name + '_extreme_lab_values'] = int(invalid.sum())
    labs.loc[invalid, 'value_std'] = np.nan
quality['lab_values_unknown_unit'] = int(labs.value_std.isna().sum())
print('Lab results with no unit before repair:', quality['lab_units_missing_before_repair'])
print('Decimal errors fixed:', sum(v for k, v in quality.items() if k.endswith('_decimal_errors_fixed')))
print('Lab results still unusable:', quality['lab_values_unknown_unit'])
display(labs.groupby('lab_name').value_std.agg(['count', 'median', 'min', 'max']).round(2))
```

In Submission 11, about 50,000 lab results (3%) were thrown away because no unit was recorded. Here the unit is inferred from the size of the value: for example, a hemoglobin of 121 must be g/L, while 12.1 must be g/dL. Labs that are only ever reported in one unit simply get that unit. Values that are exactly ten times too large (such as sodium 1,380 instead of 138) are divided by ten. Anything still outside the broad limits is set to missing.

## Severity Features

The existing lab and vital features describe the values (first, last, mean and so on). Here I add simple clinical flags: whether each measurement was low, high, or severely abnormal, and especially whether the **last** reading before discharge was still abnormal. A patient sent home with abnormal results is more likely to come back. The thresholds are common reference ranges.

```python
nan = np.nan
# (normal low, normal high, severe low, severe high)
lab_ranges = {'Sodium': (135, 145, 130, 155), 'Potassium': (3.5, 5.0, 3.0, 6.0), 'Creatinine': (nan, 1.3, nan, 2.0),
              'Bun': (nan, 20, nan, 40), 'Hemoglobin': (12, nan, 8, nan), 'Wbc': (4, 11, 2, 20),
              'Glucose': (70, 180, 54, 300), 'Albumin': (3.5, nan, 2.5, nan), 'Lactate': (nan, 2, nan, 4),
              'Troponin_I': (nan, 0.04, nan, 0.4), 'Bnp': (nan, 100, nan, 900), 'Inr': (nan, 1.5, nan, 3),
              'Hba1C': (nan, 6.5, nan, 9)}
vital_ranges = {'HR': (50, 100, 40, 130), 'SBP': (90, 160, 80, 200), 'DBP': (50, 100, 40, 120),
                'RR': (10, 22, 8, 30), 'SPO2': (92, nan, 88, nan), 'TEMP': (35.5, 38.0, 35.0, 39.0)}

def flag_severity(rows, name_column, value_column, ranges):
    limits = pd.DataFrame(ranges, index=['low', 'high', 'severe_low', 'severe_high']).T
    bounds = limits.reindex(rows[name_column]).to_numpy()
    values = rows[value_column].to_numpy()
    with np.errstate(invalid='ignore'):
        rows = rows.assign(is_low=values < bounds[:, 0], is_high=values > bounds[:, 1],
                           is_severe=(values < bounds[:, 2]) | (values > bounds[:, 3]))
    return rows.assign(is_abnormal=rows.is_low | rows.is_high)

def severity_summary(rows, name_column, prefix):
    last = rows.groupby(['encounter_id', name_column]).tail(1)
    grouped = rows.groupby('encounter_id')
    summary = pd.DataFrame({
        f'sev_{prefix}_abnormal_at_discharge': last.groupby('encounter_id').is_abnormal.sum(),
        f'sev_{prefix}_severe_at_discharge': last.groupby('encounter_id').is_severe.sum(),
        f'sev_{prefix}_n_low': grouped.is_low.sum(),
        f'sev_{prefix}_n_high': grouped.is_high.sum(),
        f'sev_{prefix}_n_severe': grouped.is_severe.sum(),
        f'sev_{prefix}_frac_abnormal': grouped.is_abnormal.mean(),
    })
    per_item = last.assign(flag=last.is_abnormal.astype(int)).pivot_table(index='encounter_id', columns=name_column, values='flag', aggfunc='max').astype(float)
    return summary.join(per_item.add_prefix(f'sev_last_abnormal_'))

lab_flags = flag_severity(labs.dropna(subset=['value_std']), 'lab_name', 'value_std', lab_ranges)
vital_flags = flag_severity(vitals.dropna(subset=['value']), 'vital_type', 'value', vital_ranges)
sev = pd.DataFrame(index=encounters.encounter_id).join(severity_summary(lab_flags, 'lab_name', 'labs'))
sev = sev.join(severity_summary(vital_flags, 'vital_type', 'vitals'))
count_columns = [c for c in sev if not c.startswith('sev_last_abnormal_') and not c.endswith('frac_abnormal')]
sev[count_columns] = sev[count_columns].fillna(0)
sev['sev_total_abnormal_at_discharge'] = sev.sev_labs_abnormal_at_discharge + sev.sev_vitals_abnormal_at_discharge
sev['sev_total_severe'] = sev.sev_labs_n_severe + sev.sev_vitals_n_severe
print(sev.shape)
sev.describe().T[['mean', 'max']].round(2).head(12)
```

## Models and Preprocessing

Numbers are median-filled with missing indicators for models that need encoded inputs (Logistic Regression); categories are one-hot encoded there. LightGBM uses native categories. Two additions for Submission 12, both computed inside each fold from the training part only:

- **Target encoding** replaces `primary_dx` with the smoothed readmission rate of that diagnosis. The training rows get their value from an inner 5-fold split, so no row ever sees its own label; validation and test rows use the rates from the whole training part of the fold.
- **Feature selection** ranks features with a small LightGBM on the training part and keeps the top ones.

LightGBM is then trained with three random seeds and the probabilities are averaged.

```python
def new_model(spec):
    parameters = copy.deepcopy(spec['params'])
    family = spec['family']
    if family == 'AdaBoost':
        depth = parameters.pop('tree_depth', 1)
        return AdaBoostClassifier(estimator=DecisionTreeClassifier(max_depth=depth,
            min_samples_leaf=100, max_features=.2, random_state=SEED), random_state=SEED, **parameters)
    if family == 'Gradient Boosting':
        return GradientBoostingClassifier(random_state=SEED, **parameters)
    if family == 'Histogram Gradient Boosting':
        return HistGradientBoostingClassifier(early_stopping=False, random_state=SEED, **parameters)
    if family == 'XGBoost':
        return XGBClassifier(objective='binary:logistic', eval_metric='auc', tree_method='hist',
                             n_jobs=JOBS, random_state=SEED, **parameters)
    if family == 'LightGBM':
        return LGBMClassifier(verbosity=-1, n_jobs=JOBS, random_state=SEED, **parameters)
    if family == 'CatBoost':
        return CatBoostClassifier(loss_function='Logloss', eval_metric='AUC', thread_count=JOBS,
            random_seed=SEED, verbose=False, allow_writing_files=False, **parameters)
    if family == 'Logistic Regression':
        return LogisticRegression(random_state=SEED, **parameters)
    raise ValueError(family)


def native_frame(frame, family):
    result = frame.copy()
    for column in result.select_dtypes(exclude='number'):
        values = result[column].fillna('MISSING').astype(str)
        result[column] = values.astype('category') if family == 'LightGBM' else values
    return result


def choose_columns(frame, target, top):
    # Feature selection is refitted inside each training fold.
    if top is None:
        return frame.columns.tolist()
    selector = LGBMClassifier(n_estimators=300, learning_rate=.03, num_leaves=7,
        min_child_samples=200, reg_lambda=30, importance_type='gain',
        verbosity=-1, n_jobs=JOBS, random_state=SEED)
    selector.fit(native_frame(frame, 'LightGBM'), target)
    importance = pd.Series(selector.feature_importances_, index=frame.columns)
    return importance.sort_values(ascending=False).head(top).index.tolist()


def target_encode(fit_keys, target, check_keys, smoothing=50):
    # Smoothed readmission rate per category, learned from the training part of the fold only.
    prior = target.mean()
    def rates(keys, labels):
        stats = pd.DataFrame({'key': keys.to_numpy(), 'y': labels.to_numpy()}).groupby('key').y.agg(['sum', 'count'])
        return (stats['sum'] + smoothing * prior) / (stats['count'] + smoothing)
    fit_encoded = np.full(len(fit_keys), prior)
    inner = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
    for inner_fit, inner_check in inner.split(fit_keys, target):
        mapping = rates(fit_keys.iloc[inner_fit], target.iloc[inner_fit])
        fit_encoded[inner_check] = fit_keys.iloc[inner_check].map(mapping).fillna(prior).to_numpy()
    check_encoded = check_keys.map(rates(fit_keys, target)).fillna(prior).to_numpy()
    return fit_encoded, check_encoded


def fit_predict(spec, fitting, target, checking):
    fitting, checking = fitting.copy(), checking.copy()
    for column in spec.get('target_encode', []):
        fit_keys = fitting[column].fillna('MISSING').astype(str)
        check_keys = checking[column].fillna('MISSING').astype(str)
        fitting[column + '_te'], checking[column + '_te'] = target_encode(fit_keys, target, check_keys)
        fitting, checking = fitting.drop(columns=column), checking.drop(columns=column)
    columns = choose_columns(fitting, target, spec.get('select_top'))
    fitting, checking = fitting[columns], checking[columns]
    model = new_model(spec)
    family = spec['family']
    if family in ['LightGBM', 'CatBoost']:
        fit_values, check_values = native_frame(fitting, family), native_frame(checking, family)
        if family == 'CatBoost':
            model.set_params(cat_features=fitting.select_dtypes(exclude='number').columns.tolist())
        transformer = None
    else:
        numeric = fitting.select_dtypes('number').columns.tolist()
        categorical = fitting.columns.difference(numeric).tolist()
        transformer = ColumnTransformer([
            ('number', Pipeline([('fill', SimpleImputer(strategy='median', add_indicator=True, keep_empty_features=True)),
                                 ('scale', StandardScaler())]), numeric),
            ('category', Pipeline([('fill', SimpleImputer(strategy='constant', fill_value='MISSING')),
                                   ('encode', OneHotEncoder(handle_unknown='ignore', sparse_output=False, dtype=np.float32))]), categorical),
        ], sparse_threshold=0)
        fit_values = transformer.fit_transform(fitting).astype(np.float32)
        check_values = transformer.transform(checking).astype(np.float32)
    seeds = spec.get('seeds', [SEED])
    check_probability, fit_probability = np.zeros(len(check_values)), np.zeros(len(fit_values))
    for seed in seeds:
        model = new_model(spec)
        if family == 'LightGBM':
            model.set_params(random_state=seed)
        if family == 'CatBoost':
            model.set_params(cat_features=fitting.select_dtypes(exclude='number').columns.tolist())
        model.fit(fit_values, target)
        check_probability += model.predict_proba(check_values)[:, 1] / len(seeds)
        fit_probability += model.predict_proba(fit_values)[:, 1] / len(seeds)
    return check_probability, fit_probability, model, transformer, columns
```

## Trials

LightGBM and Logistic Regression are unchanged from Submission 13. The new trial is CatBoost on the compact view (the same view as LightGBM):

- `depth = 4`: shallow trees, in line with the small trees that worked best for LightGBM.
- `learning_rate = 0.03` with `iterations = 1500`.
- `l2_leaf_reg = 10`: a stronger penalty on leaf values than the default (3), to limit overfitting.
- `rsm = 0.3`: each split looks at a random 30% of the features (like `colsample_bytree` in LightGBM).
- The category columns are passed to CatBoost as categories, so no one-hot encoding or manual target encoding is needed.

```python
TRIALS = {
    'LightGBM improved': {'family': 'LightGBM', 'params': {'n_estimators': 3000, 'learning_rate': 0.01, 'num_leaves': 4, 'min_child_samples': 400, 'reg_lambda': 30, 'colsample_bytree': 0.3, 'subsample': 0.8, 'subsample_freq': 1}, 'view': 'compact',
                          'target_encode': ['primary_dx'], 'select_top': 120, 'seeds': [42, 7, 2026]},
    'CatBoost': {'family': 'CatBoost', 'params': {'iterations': 1500, 'learning_rate': 0.03, 'depth': 4, 'l2_leaf_reg': 10, 'rsm': 0.3},
                 'view': 'compact', 'select_top': None},
    'Logistic Regression': {'family': 'Logistic Regression', 'params': {'penalty': 'l1', 'C': 0.1, 'solver': 'liblinear', 'max_iter': 2000}, 'view': 'full', 'select_top': None},
}
# Blend weights fixed in advance. The first one is exactly the Submission 13 blend.
BLENDS = {
    'Blend 80/20 (Submission 13)': {'LightGBM improved': 0.8, 'Logistic Regression': 0.2},
    'Blend 60/20/20':              {'LightGBM improved': 0.6, 'CatBoost': 0.2, 'Logistic Regression': 0.2},
    'Blend 50/30/20':              {'LightGBM improved': 0.5, 'CatBoost': 0.3, 'Logistic Regression': 0.2},
    'Blend 40/40/20':              {'LightGBM improved': 0.4, 'CatBoost': 0.4, 'Logistic Regression': 0.2},
}
SUB11_BLEND_CV = 0.75176      # Submission 11: 70/30 blend (reference from the previous notebook)
SUB11_LIGHTGBM_CV = 0.75106   # Submission 11: LightGBM small trees
SUB12_BLEND_CV = 0.75268     # Submission 12: 80/20 blend with L2 Logistic Regression
SUB12_LOGISTIC_CV = 0.74741  # Submission 12: L2 Logistic Regression
PREVIOUS_CV = SUB12_BLEND_CV
```

## Cross-Validation

The same five stratified folds with seed 42 are used throughout. Mean ROC-AUC selects the trial. Accuracy uses a 0.5 threshold and is secondary because readmissions are uncommon. Reusing cross-validation to select trials produces development scores; an unseen leaderboard result can differ.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
rows, oof = [], {}
for name, spec in TRIALS.items():
    frame = views[spec['view']].loc[y.index]
    probabilities = pd.Series(index=y.index, dtype=float)
    fold_scores, training_scores = [], []
    started = perf_counter()
    with threadpool_limits(limits=JOBS):
        for fold, (fit_positions, check_positions) in enumerate(cv.split(frame, y), start=1):
            fit_rows, check_rows = frame.iloc[fit_positions], frame.iloc[check_positions]
            probability, train_probability, _, _, _ = fit_predict(spec, fit_rows, y.iloc[fit_positions], check_rows)
            probabilities.iloc[check_positions] = probability
            fold_scores.append(roc_auc_score(y.iloc[check_positions], probability))
            training_scores.append(roc_auc_score(y.iloc[fit_positions], train_probability))
            print(f'{name} | fold {fold}: {fold_scores[-1]:.5f}', flush=True)
    predicted = probabilities.ge(.5)
    row = {'Trial': name, 'Model': spec['family'], 'View': spec['view'],
           'Features': min(len(frame.columns), spec.get('select_top') or len(frame.columns)),
           'ROC-AUC': float(np.mean(fold_scores)), 'Std': float(np.std(fold_scores)),
           'Train ROC-AUC': float(np.mean(training_scores)),
           'Accuracy': accuracy_score(y, predicted), 'Precision': precision_score(y, predicted, zero_division=0),
           'Recall': recall_score(y, predicted, zero_division=0), 'F1': f1_score(y, predicted, zero_division=0),
           'Average precision': average_precision_score(y, probabilities),
           'Seconds': round(perf_counter() - started, 1)}
    row['Gap'] = row['Train ROC-AUC'] - row['ROC-AUC']
    rows.append(row)
    oof[name] = probabilities
    pd.DataFrame(rows).to_csv(OUTPUT_DIR / 'model_scores.csv', index=False)
    pd.DataFrame({'fold': range(1, 6), 'ROC-AUC': fold_scores}).to_csv(OUTPUT_DIR / (re.sub(r'[^a-zA-Z0-9]+', '_', name) + '_folds.csv'), index=False)
    print(f'{name}: mean CV AUC {row["ROC-AUC"]:.5f} +/- {row["Std"]:.5f}', flush=True)

summary = pd.DataFrame(rows).sort_values('ROC-AUC', ascending=False).reset_index(drop=True)
display(summary[['Trial', 'Model', 'View', 'Features', 'ROC-AUC', 'Std', 'Accuracy', 'Train ROC-AUC', 'Gap']].round(5))
```

## Blending: two models vs three models

Each model's predictions are turned into ranks (0 = lowest risk, 1 = highest risk) before blending, because the models give probabilities on different scales. The blend is a weighted average of the ranks, using the weights fixed above. This is weighted rank blending, not stacking: no second model learns the weights.

All blends use the same out-of-fold predictions, so they are scored on the same five folds as the single models. Choosing the best of four weight settings makes its CV very slightly optimistic, so the fold-by-fold table is shown as well: a real improvement should appear in most folds, not just in the mean.

```python
def to_rank(values):
    return rankdata(values) / len(values)

def fold_scores(predictions):
    return np.array([roc_auc_score(y.iloc[check], predictions.iloc[check]) for _, check in cv.split(predictions, y)])

ranks = {name: to_rank(oof[name]) for name in TRIALS}
display(pd.DataFrame({a: {b: np.corrcoef(ranks[a], ranks[b])[0, 1] for b in TRIALS} for a in TRIALS}).round(3)
        .rename_axis('Rank correlation'))

blend_rows, folds_table = [], {'Fold': range(1, 6), 'LightGBM improved': fold_scores(oof['LightGBM improved']).round(4)}
for blend_name, weights in BLENDS.items():
    blend = pd.Series(sum(weight * ranks[name] for name, weight in weights.items()), index=y.index)
    blend_folds = fold_scores(blend)
    oof[blend_name] = blend
    folds_table[blend_name] = blend_folds.round(4)
    blend_rows.append({'Trial': blend_name, 'Model': ' + '.join(weights), 'View': 'compact + full',
                       'Features': np.nan, 'ROC-AUC': float(blend_folds.mean()), 'Std': float(blend_folds.std()),
                       'Accuracy': float('nan')})
display(pd.DataFrame(folds_table))
display(pd.DataFrame(blend_rows)[['Trial', 'ROC-AUC', 'Std']].round(5))

summary = pd.concat([summary, pd.DataFrame(blend_rows)]).sort_values('ROC-AUC', ascending=False).reset_index(drop=True)
```

## Final Model and Submission

The final prediction follows exactly the same steps as the validation. For each of the five folds, each model in the selected blend is trained on that fold's training part (LightGBM with its own target encoding, feature selection and three seeds) and it scores the test stays. The five fold predictions are averaged per model, turned into ranks, and combined with the selected blend weights. If a single model scores higher than every blend, that model is used instead.

```python
best = summary.iloc[0]
best_name = best.Trial
weights = BLENDS[best_name] if best_name in BLENDS else {best_name: 1.0}
best_spec = {name: copy.deepcopy(TRIALS[name]) for name in weights}
model_predictions = {}
selected_columns = []
submission_folds = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
with threadpool_limits(limits=JOBS):
    for name in weights:
        frame = views[TRIALS[name]['view']]
        labeled, unlabeled = frame.loc[y.index], frame.loc[test.encounter_id]
        fold_predictions = []
        for fit_positions, _ in submission_folds.split(labeled, y):
            probability, _, _, _, columns = fit_predict(
                TRIALS[name], labeled.iloc[fit_positions], y.iloc[fit_positions], unlabeled)
            fold_predictions.append(probability)
        model_predictions[name] = np.mean(fold_predictions, axis=0)
        selected_columns = list(dict.fromkeys(selected_columns + columns))
        print(name, 'done', flush=True)
if len(weights) == 1:
    test_probability = model_predictions[best_name]
else:
    test_probability = sum(weight * to_rank(model_predictions[name]) for name, weight in weights.items())
submission = pd.DataFrame({'encounter_id': test.encounter_id.to_numpy(), 'readmit_30d': test_probability})
sample = pd.read_csv(DATA_DIR / 'sample_submission.csv')
submission = submission.set_index('encounter_id').loc[sample.encounter_id].reset_index()
assert submission.encounter_id.equals(sample.encounter_id)
assert submission.readmit_30d.between(0, 1).all()
submission.to_csv(OUTPUT_DIR / 'submission.csv', index=False)
pd.DataFrame({'feature': selected_columns}).to_csv(OUTPUT_DIR / 'final_columns.csv', index=False)
selected = {'stage': STAGE, 'stage_name': STAGE_NAME, 'trial': best_name,
            'spec': best_spec, 'cv_roc_auc': float(best['ROC-AUC']),
            'cv_std': float(best.Std), 'accuracy': float(best.Accuracy),
            'previous_cv': PREVIOUS_CV, 'base_view': BASE_VIEW,
            'new_public_score': None,
            'validation': 'Five-fold stratified CV, shuffle=True, seed=42. CV also selects trials; it is a development estimate.',
            'blend_weights': weights,
            'prediction': 'Per model: average of five fold models (LightGBM: three seeds each); weighted rank blend.'}
(OUTPUT_DIR / 'selected_model.json').write_text(json.dumps(selected, indent=2))
(OUTPUT_DIR / 'data_quality.json').write_text(json.dumps(quality, indent=2))
print('Selected:', best_name, weights)
print('Mean CV ROC-AUC:', f'{best["ROC-AUC"]:.5f}')
print('Input features:', len(selected_columns))
print('Submission rows:', len(submission))
print('Saved: submission.csv')
display(submission.head())
```
